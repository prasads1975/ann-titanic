```python
from tensorflow.keras.models import load_model
import pickle
import pandas as pd
import tensorflow as tf
```

```python
with open('classificiation_label_encoder.pkl','rb') as file:
    label = pickle.load(file)
```

```python
with open('classificiation_one_hot_encoder.pkl','rb') as file:
    ohe = pickle.load(file)
```

```python
with open('classificiation_scaler.pkl','rb') as file:
    scaler = pickle.load(file)
```

```python
model = load_model('classification_model.h5')
```

```python
user= {'Pclass':1,'Sex':'male','SibSp':2,'Parch':1,'Fare':76,'Embarked':'Cheboug'}
```

```python
user_df = pd.DataFrame([user])
```

```python
user_df
```

```python
user_df['Sex'] = label.transform(user_df['Sex'])
```

```python
user_df
```

```python
embarked = ohe.transform(user_df[['Embarked']])
```

```python
embarked_df = pd.DataFrame(embarked,columns=ohe.get_feature_names_out())
```

```python
embarked_df
```

```python
user_df = pd.concat([user_df.drop(columns=['Embarked']),embarked_df],axis=1)
```

```python
user_df
```

```python
user_df[['Pclass','SibSp','Parch','Fare']] =  scaler.transform(user_df[['Pclass','SibSp','Parch','Fare']])
```

```python
user_df
```

```python
y = model.predict(user_df)
```

```python
y
```

```python
y = y[0][0]
```

```python
y
```

```python
if y > 0.5:
    print('Passenger will survive')
else:
    print('Passenger wont survive')
```
