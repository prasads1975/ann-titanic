## Feature Encoding

```python
## Encode Sex --> LabelEncoder
from sklearn.preprocessing import LabelEncoder
```

```python
label=LabelEncoder()
```

```python
data['Sex'] = label.fit_transform(data['Sex'])
```

```python
data.head()
```

```python
## Encode Embarked --> OneHotEncoder
from sklearn.preprocessing import OneHotEncoder
```

```python
ohe = OneHotEncoder(sparse_output=False)
```

```python
embarked = ohe.fit_transform(data[['Embarked']])
```

```python
## create data frame of the encoded embarked data
embarked_df = pd.DataFrame(embarked,columns=ohe.get_feature_names_out())
```

```python
embarked_df
```

```python
embarked_df.shape
```

```python
embarked_df.isnull().sum()
```

```python
## join the data frames
data = pd.concat([data.drop(columns=['Embarked']),embarked_df],axis=1)
```

## Train / Test / Val Split

```python
## Input Columns - Drop the result colum
x=data.drop(columns=['Survived'])
```

```python
x.head()
```

```python
x.isnull().sum()
```

```python
## Output Column
y=data['Survived']
```

```python
y.head()
```

```python
y.isnull().sum()
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=20,random_state=42)
```

```python
## Validation data
x_train,x_val,y_train,y_val = train_test_split(x_train,y_train,test_size=0.2,random_state=42)
```

```python
x_val.shape
```

```python
x_train.isnull().sum()
```

```python
y_train.isnull().sum()
```

```python
import tensorflow
```

```python
# Deep learning is sequential hence the Sequential model
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense # Hidden layer
# stop early if the loss function or acurracy is not changing even after changing epochs, this remain same
from tensorflow.keras.callbacks import EarlyStopping 
```

```python
## Input nodes should be same as the number of inpur features (columns)
(x_train.shape[1],)
```

```python
#model = Sequential([
# Dense(64,input_shape=(x_train.shape[1],),activation='relu'), #First Hidden layer
# Dense(32,activation='relu'), # second hidden layer
# Dense(1,activation='sigmoid') # output layer, as this is binary classification activation function on output is sigmoid
# ])
```

```python
model = Sequential([Dense(64,input_shape=(x_train.shape[1],),activation='relu'),
        Dense(32,activation='relu'), 
        Dense(1,activation='sigmoid')]   )
```

```python
model.summary()
```

## Model Compilation --> Optimizer, Loss, Metrics

```python
opt = tensorflow.keras.optimizers.Adam(learning_rate=0.01)
```

```python
## loss function is binary_crossentropy as this a binary classification problem
model.compile(optimizer=opt,loss='binary_crossentropy',metrics=['accuracy'])
```

```python
# monitor -> validation loss
# patience -> when to stop (epoch)
stopping_callback = EarlyStopping(monitor='val_loss',patience=5,restore_best_weights=True)
```

```python
model.fit(x_train,y_train,validation_data=(x_val,y_val),epochs=100,callbacks=[stopping_callback])
```

```python
# store in h5 format to maintain the original formats of the weights
model.save('classification_model.h5')
```
