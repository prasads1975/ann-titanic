### No Ecoding required as all columns are numeric only Scaling

```python
from sklearn.preprocessing import StandardScaler
scale = StandardScaler()
```

```python
x = data.drop(columns=['cognitive_load'])
y = data['cognitive_load']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y, test_size=0.2,random_state=42)
```

```python
x_train.shape
```

```python
y_train.shape
```

```python
x.columns
```

```python
x_train[['hours_coding', 'coffee_intake_mg', 'distractions', 'sleep_hours',
       'commits', 'bugs_reported', 'ai_usage_hours']] = scale.fit_transform(x_train[['hours_coding', 'coffee_intake_mg', 'distractions', 'sleep_hours',
       'commits', 'bugs_reported', 'ai_usage_hours']])
```

```python
x_train
```

```python
x_test[['hours_coding', 'coffee_intake_mg', 'distractions', 'sleep_hours',
       'commits', 'bugs_reported', 'ai_usage_hours']] = scale.transform(x_test[['hours_coding', 'coffee_intake_mg', 'distractions', 'sleep_hours',
       'commits', 'bugs_reported', 'ai_usage_hours']])
```

```python
x_test
```

```python
(x_train.shape[1],)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.callbacks import EarlyStopping
```

```python
model = Sequential([
    Dense(64,input_shape=(x_train.shape[1],),activation='linear'),
    Dense(32,activation='linear'),
    Dense(1,activation='linear')
])
```

```python
model.summary()
```

```python
model.compile(optimizer='adam',metrics=['mae'],loss='mse')
```

```python
model.fit(x_train,y_train,epochs=100)
```
